```python
from goodrich.ch09.unsorted_priority_queue import UnsortedPriorityQueue
```

```python
pq = UnsortedPriorityQueue()

pq.add(3, "tarea C")
pq.add(1, "tarea A")
pq.add(2, "tarea B")

print(pq.min())         # (1, 'tarea A')
print(pq.remove_min())  # (1, 'tarea A')
print(pq.min())         # (2, 'tarea B')
```

```python
class Empty(Exception):
    pass

class Stack_pq():
    def __init__(self):
        self._pq = UnsortedPriorityQueue()
        self._count = 0

    def push(self, item):
        self._pq.add(self._count, item)
        self._count -= 1

    def pop(self):
        return self._pq.remove_min()[1]

    def top(self):
        return self._pq.min()[1]

    def is_empty(self):
        if len(self._pq) == 0:
            raise Empty("El Stack está vacío")

    def __len__(self):
        return len(self._pq)
```
